```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
rng = np.random.default_rng(1)

n = 500
x1 = rng.normal(size=n)
x2 = rng.normal(size=n)

y = ((x1 > 0) & (x2 > 0)).astype(int)

noise = rng.choice(n, size=50, replace=False)
y[noise] = 1 - y[noise]

X = pd.DataFrame({
    "x1": x1,
    "x2": x2,
    #"x1*x2": x1 * x2
})

X_train, X_test, y_train, y_test = \
    train_test_split(X, y, test_size=0.2, random_state=1)

logistic = LogisticRegression()
logistic.fit(X_train, y_train)

print("coef =", logistic.coef_)
print("train_score =", logistic.score(X_train, y_train))
print("test_score =", logistic.score(X_test, y_test))

# score of a baseline model: 0.75 (when no noise)
```

```python
Tree1 = DecisionTreeClassifier(max_depth=1)
Tree2 = DecisionTreeClassifier(max_depth=2)
Tree3 = DecisionTreeClassifier(max_depth=None)

Tree1.fit(X_train, y_train)
Tree2.fit(X_train, y_train)
Tree3.fit(X_train, y_train)

print("Tree1 (max_depth=1)")
print("train_score =", Tree1.score(X_train, y_train))
print("test_score  =", Tree1.score(X_test, y_test))

print("\nTree2 (max_depth=2)")
print("train_score =", Tree2.score(X_train, y_train))
print("test_score  =", Tree2.score(X_test, y_test))

print("\nTree3 (max_depth=None)")
print("train_score =", Tree3.score(X_train, y_train))
print("test_score  =", Tree3.score(X_test, y_test))
```

```python
from sklearn.tree import export_text

print("Tree2")
print(export_text(Tree2, feature_names=["x1", "x2"]))

print("\nTree3")
print(export_text(Tree3, feature_names=["x1", "x2"]))
```
